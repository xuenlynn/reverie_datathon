```python
# Before vs After: Validate the Cleaning Pipeline
# =========================================
# This section re-imports raw and processed data independently to confirm
# split_data.py's cleaning/feature engineering behaved as expected.

df_raw = pd.read_csv('../data/raw/marketing_campaign.csv', sep='\t')
df_train = pd.read_csv('../data/processed/train.csv')
df_val = pd.read_csv('../data/processed/val.csv')
df_test = pd.read_csv('../data/processed/test.csv')

# Recombine splits to compare against the full raw dataset
df_processed = pd.concat([df_train, df_val, df_test], ignore_index=True)

print(f"Raw shape:        {df_raw.shape}")
print(f"Processed shape:  {df_processed.shape}")
print(f"Rows removed:     {df_raw.shape[0] - df_processed.shape[0]} "
      f"({(df_raw.shape[0] - df_processed.shape[0]) / df_raw.shape[0] * 100:.1f}%)")
```
